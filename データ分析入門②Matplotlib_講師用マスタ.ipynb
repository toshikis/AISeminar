## <font color=#0e89b2>1.1 散布図の作成</font>

【説明】<br>
散布図は、ご存知の通り、体重と身長、英語の点数と数学の点数など、2つのペアのデータを1つの点としてプロットしたグラフです。<br>
2つのデータの相関や、分布の傾向を見る際に便利です。<br>
<br>
【実装コード】<br>
`plt.scatter(x, y)`

```python
#####写経#####
#散布図の作成
#散布図の作成
#データの準備
x = np.random.random(30) #x軸のデータ
y = np.random.random(30) #y軸のデータ

#キャンバスの描画
plt.figure(figsize = (10,6)) #キャンパスの大きさの指定
plt.title("散布図") #グラフのタイトル
plt.xlabel("x") #xの座標名
plt.ylabel("y") #yの座標名
plt.grid(True) #グリッド線の表示

#グラフの描画
plt.scatter(x,y)
```

ランダムな値を生成したので、グラフ上も規則性なく点がプロットされているはずです。<br>
<br>
ちなみに上では丁寧にキャンバスの大きさやグラフタイトルなどをいちいち指定してコードを書いていますが、<br>
ミニマムではこれだけ（コード1行)だけでも描画することができます。

```python
#####写経#####
#散布図の作成
plt.scatter([4,3,6], [2,4,5])
```

## <font color=#0e89b2>1.2 折れ線グラフの作成</font>

【説明】<br>
折れ線グラフももはや説明不要だとは思いますが、ある一つの変数の値の変化を見る時に使用するグラフです。<br>
<br>
【実装コード】<br>
`plt.plot(x, y)`

```python
#####写経#####
#折れ線グラフの作成
#データの準備
x = np.arange(30) #x軸のデータ。arangeは連番を生成する関数
y = np.random.randn(30).cumsum() #y軸のデータ。cumsumは累積合計を取得する関数

#キャンバスの描画
plt.figure(figsize = (10,6)) #グラフの大きさの指定
plt.title("折れ線グラフ") #グラフのタイトル
plt.xlabel("x") #xの座標名
plt.ylabel("y") #yの座標名
plt.grid(True) #グリッド線の表示

#グラフの描画
plt.plot(x,y)
```

### <font color=#0e89b2>練習問題1.1：花弁の長さのヒストグラムの作成</font>

花弁の長さをヒストグラムで集計してください。<br>

# <font color=#0e89b2>2. 複数グラフの表示</font>

## <font color=#0e89b2>2.1 Pyplotスタイルとオブジェクト指向スタイル</font>

ここで、matplotlibの新しい書き方を学ぶ必要があります。<br>
実はmatplotlibの書き方には、大きく2つの流儀があります。<br>

* Pyplotスタイル
 * 簡単な書き方
 * `plt.なになに`で全部済ませる
* オブジェクト指向スタイル
 * 丁寧な書き方
 * `fig.なになに`と`ax.なになに`を書き分ける

<br>

ではこの`fig`や`ax`とはどういう意味でしょうか？

image.png

`fig`というのは上の図の中の`Figure`、`ax`というのは上の図の中の`Axes`を指しています。

image.png

図のように、`Figure`はダッシュボード全体、`Axes`は個々のグラフのイメージです。<br>
<br>
Pyplotスタイルとオブジェクト指向スタイルの出来ることの大きな違いは、<br>
複数のグラフを並べて表示できるかどうかです。<br>

image.png

`plt.なになに`で全部済ませるMATLABスタイルでは、<br>
「ダッシュボードの中に良い感じにこの散布図描いといてー！」的な雑な指示だったわけです。<br>
これだと指示が曖昧なので、複数グラフの描画には対応できませんでした。<br>
<br>
一方オブジェクト指向スタイルでは、<br>
「まずダッシュボード全体を作って」<br>
「ダッシュボードの中に2行×3列で計6個のグラフを描くスペースをそれぞれ作って」<br>

今回はモンテカルロ法を用いて、円周率を求めてみましょう。

##<font color=#0e89b2>総合問題1-1：乱数の発生</font>

0から1の間の一様分布に従う乱数10,000個を2組発生させましょう。<br>
ヒント：<br>
・一様乱数とは、ある数から数まで等確率で発生する乱数のことです。<br>
・一様乱数を発生させるには、np.random.uniformを使います。

```python
#0から1の間の一様分布に従う乱数10,000個を2組発生させましょう
#サンプル数の設定


#1組目：乱数x


#2組目：乱数y
```

解答

```python
#0から1の間の一様分布に従う乱数10,000個を2組発生させましょう
#サンプル数の設定
N=10000

#1組目：乱数x
x = np.random.uniform(0.0, 1.0, N)

#2組目：乱数y
y = np.random.uniform(0.0, 1.0, N)
```

##<font color=#0e89b2>総合問題1-2：点が円の内部にあるか外部にあるかの判定</font>

図のような半径1の円と長さ1の正方形を考えます。<br>
1-1の（x、y）の組み合わせのうち、円の内部に入る点は何個ありますか？<br>
<br>
ヒント：「円の内部に入る」ことは、「xとyの二乗の和の平方根(つまり原点からの距離)が1未満であること」と言い換えられます。

```python
#円の内部に入る点は何個ありますか？
#円の内部に入ったxとyを格納するリストinside_x, inside_yを用意


#円の外部に出たxとyを格納するリストoutside_x, outside_yを用意


#「xとyの二乗の平方根」が1未満であればリストA、1以上であればリストBに振り分け
```

解答

```python
count = 0
for i in range(0, N):
  distance = np.sqrt(x[i]**2 + y[i]**2)
  if distance < 1:
    count += 1
count
```

##<font color=#0e89b2>総合問題1-3：円と点の可視化</font>

総合問題1-2で円の内部に入った点を青色、外部の点を赤色で、プロットしてみてください。<br>

```python
#データの準備
#円の内部に入ったxとyを格納するリストinside_x, inside_yを用意
inside_x = []
inside_y = []

#円の外部に出たxとyを格納するリストoutside_x, outside_yを用意
outside_x = []
outside_y = []

#「xとyの二乗の平方根」が1未満であればリストA、1以上であればリストBに振り分け
count = 0
for i in range(1000):
  distance = np.sqrt(x[i]**2 + y[i]**2)
  if distance < 1:
    count += 1
    inside_x.append(x[i])
    inside_y.append(y[i])
  else:
    outside_x.append(x[i])
    outside_y.append(y[i])
```

```python
#円の内部に入る点を青色、外部の点を赤色でプロット
plt.figure(figsize = (6,6))
plt.scatter(inside_x, inside_y, color="blue", s=10)
plt.scatter(outside_x, outside_y, color="red", s=10)
```

##<font color=#0e89b2>総合問題1-4：円周率の算出</font>

上の図の四分円の面積はπ/4、正方形の面積は1です。<br>
<br>
仮に四分円の中に入っている点の個数が15個、正方形の中に入っている点の個数が100個であれば、<br>
π/4：1=15：100<br>
π：1=30：100<br>
π=3<br>
と計算できます。
<br>
この考え方を利用して、円周率を求めてみてください。

```python
#円周率を求める
4*count/N
```

また、アーティストや楽曲の情報を取得するための関数も、今回は時間の都合上準備しておきます。

```python
#任意のアーティスト名から、そのアーティストのIDを返す
def getIdByArtist(artist_name):
    results = spotify.search(q="artist:" + artist_name, type="artist")
    items = results["artists"]["items"]
    artist = items[0]
    artist_id = artist["id"]
    return artist_id
```

```python
#任意のアーティスト名から、そのアーティストの人気曲TOP10を返す
def getTopSongs(artist_name):
    num = 10
    track_name_ls = []
    track_id_ls = []
    try:
        search_id = getIdByArtist(artist_name)
        top_tracks = spotify.artist_top_tracks(search_id, country="JP")["tracks"]

        for i in range(num):
          track_name_ls.append(top_tracks[i]["name"])
          track_id_ls.append(top_tracks[i]["id"])

        df = pd.DataFrame(columns=["track_name","track_id"])
        df["track_name"] = track_name_ls
        df["track_id"] = track_id_ls
        return df
    except IndexError:
        print("IndexError")
    except AttributeError:
        print("AttributeError")
```

```python
#任意のアーティスト名から、関連アーティスト名を返す
def getRelatedArtists(artist_name):
  name_ls = []
  id_ls = []

  artist_id = getIdByArtist(artist_name)
  related_artists = spotify.artist_related_artists(artist_id)
  for i in spotify.artist_related_artists(artist_id)['artists']:
    name_ls.append(i['name'])
    id_ls.append(i['id'])

  df = pd.DataFrame(columns=["artist_name","artist_id"])
  df["artist_name"] = name_ls
  df["artist_id"] =id_ls
  return df
```

```python
def getTrackFeatures(id):
    meta = spotify.track(id)
    features = spotify.audio_features(id)
    df = pd.DataFrame(columns=["name", "artist", "popularity", "acousticness", "danceability", "energy", "instrumentalness", "mode", "liveness", "loudness", "speechiness", "tempo", "time_signature", "valence"])

    #meta
    df["name"] = [meta['name']]
    df["artist"] = [meta['album']['artists'][0]['name']]
    df["popularity"] = [meta['popularity']]
    #features
    df["acousticness"] = features[0]['acousticness']
    df["danceability"] = [features[0]['danceability']]
    df["energy"] = [features[0]['energy']]
    df["instrumentalness"] = [features[0]['instrumentalness']]
    df["mode"] = [features[0]['mode']]
    df["liveness"] = [features[0]['liveness']]
    df["loudness"] = [features[0]['loudness']]
    df["speechiness"] = [features[0]['speechiness']]
    df["tempo"] = [features[0]['tempo']]
    df["time_signature"] = [features[0]['time_signature']]
    df["valence"] = [features[0]['valence']]
    return df
```

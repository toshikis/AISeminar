## <font color=#0e89b2>1.3 「AIのためのPython入門」のゴール</font>

今回は0からPythonを学んでいくのですが、<br>
せっかくAIの講座ですから、ただ文法を学ぶだけでなく、同時にシミュレーションモデルを作っていきます。<br>
<br>
最終的に完成するシミュレーションモデルはこちらです。<br>
`Shift+Enter`で実行してみてください。(実行には数十秒ほどかかります)

```python
#このコードは難読化されています
! pip install japanize_matplotlib
import numpy as np
from random import random
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import japanize_matplotlib
from IPython.display import HTML
#このコードは難読化されています
def abc(population, qwe, asd):
    zxc = [{'vbn': np.random.randint(qwe), 'bnm': np.random.randint(qwe), 'nmo': np.random.choice([-1, 0, 1]),
            'mop': np.random.choice([-1, 0, 1]), 'poi': False, 'iuy': random() < asd, 'uyt': 0} for _ in range(population)]
    zxc[0].update({'poi': True, 'iuy': False})
    return zxc

def defg(zxc, qwe):
    for lmn in zxc:
        lmn['nmo'] *= -1 if lmn['vbn'] + lmn['nmo'] <= 0 or lmn['vbn'] + lmn['nmo'] >= qwe else 1
        lmn['mop'] *= -1 if lmn['bnm'] + lmn['mop'] <= 0 or lmn['bnm'] + lmn['mop'] >= qwe else 1
        lmn['vbn'] += lmn['nmo']
        lmn['bnm'] += lmn['mop']

def hij(zxc):
    for lmn in zxc:
        if lmn['poi'] and lmn['uyt'] >= 6: lmn.update({'poi': False, 'uyt': 0})
        elif lmn['poi']: lmn['uyt'] += 1

def klm(lmn1, lmn2):
    return abs(lmn1['vbn'] - lmn2['vbn']) <= 2 and abs(lmn1['bnm'] - lmn2['bnm']) <= 2

def nop(lmn1, lmn2, ghj, jkl):
    return ghj if lmn1['iuy'] or lmn2['iuy'] else jkl

def qrs(zxc, ghj, jkl):
    for lmn in zxc:
        if lmn['poi']:
            for fgh in [f for f in zxc if klm(lmn, f) and not f['poi']]:
                if random() < nop(lmn, fgh, ghj, jkl):
                    fgh.update({'poi': True, 'uyt': 0})

def tuv(wxy, zxc, ghj, jkl, qwe, yza):
    defg(zxc, qwe)
    hij(zxc)
    qrs(zxc, ghj, jkl)
    abc = [ljk['poi'] for ljk in zxc].count(True)
    efg.append(abc)
    rst, uvw = zip(*[(lmn['vbn'], lmn['bnm']) for lmn in zxc if not lmn['poi']]), zip(*[(lmn['vbn'], lmn['bnm']) for lmn in zxc if lmn['poi']])
    hik.clear()
    hik.set(xlim=(0, qwe), ylim=(0, qwe), title=f'{wxy}日目')
    if rst: hik.scatter(*rst, color='blue', s=10, alpha=0.5)
    if uvw: hik.scatter(*uvw, color='red', s=10, alpha=0.5)
    lmn.clear()
    lmn.set(xlim=(0, yza), ylim=(0, pop), title='感染拡大状況')
    lmn.plot(efg, color='red', alpha=0.5)

pop, qwe, asd, ghj, jkl, yza, efg = 1000, 100, 0.5, 0.1, 0.8, 150, []
fig, (hik, lmn) = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle("マスク着用率：50%")
zxc = abc(pop, qwe, asd)
ani = FuncAnimation(fig, tuv, frames=range(yza), fargs=(zxc, ghj, jkl, qwe, yza), interval=100)
plt.close(fig)
HTML(ani.to_jshtml())
```

これは、新型コロナウィルスの感染拡大状況をシミュレーションしたモデルです。<br>
マスクの着用率が変わることによって、どのように感染拡大規模・スピードが変化するかをシミュレーションします。<br>


---
◆シミュレーションの条件
* 人口は1,000人。
* それぞれが100×100マスのグリッド（街）の中を動き回る。
 * 上下左右斜めのいずれかの方向に直線的に進み、空間の端に到達したら反転してまた進む。
 * 進むスピードは1日に1マスずつ。
* 最初に1人だけ新型コロナウィルスに罹患している。
* 人同士が2マス以内の距離に近づくと、以下の確率で感染する。
 * マスクを着用している場合：10%
 * マスクを着用していない場合：80%
* 新型コロナウィルスに罹患しても、7日間経つと治癒する。（再感染の可能性はあり）
* 300日分シミュレーションを行う

```python
price = 800
tax = 0.1
price*(1+tax)
```


> この現象は、浮動小数点数の計算における精度の問題によるものです。コンピューターでは、浮動小数点数（実数）は2進数（バイナリ）で表現されます。しかし、2進数で表現する際には、10進数の小数を正確に表現できないケースがあります。特に、0.1のような簡単な10進数の小数でも、2進数では無限に続く循環小数となってしまうため、完全に正確な値を表現できません。

ちなみにこの問題を解決するためには、以下のようにround()関数を使用して、<br>
計算結果を必要な小数点以下の桁数で丸めることが一般的です。

```python
round(price*(1+tax))
```

 ### <font color=#0e89b2>(3) ループ処理</font>

for文を使うと、リストの要素を順番に一つずつ取り出すことができます。

image.png

```python
languages = ["Python", "PHP", "Ruby", "Java"]
for language in languages:
  print(language)
```

`sum()`関数を使わず、for文を使って合計値を算出してみます。

```python
scores_sum = 0
scores = [76, 45, 89]
for score in scores:
  scores_sum += score
scores_sum
```

またfor文は、`range()`関数と組み合わせて使われることが非常に多いです。<br>
実際に挙動を確認してみましょう。

```python
for i in range(3):
  print(i)
```

このように、`range(3)`を生成してfor文で順番に取り出すと、<br>
0,1,2の数字が順番に現れます。(3は含まれないことに注意です）<br>
<br>
範囲を指定することもできます。

```python
for i in range(10, 13):
  print(i)
```

`range()`関数を使うと、色々な処理が可能になります。

```python
for i in range(5):
  print("5回繰り返して出力")
```

```python
languages = ["Python", "PHP", "Ruby", "Java"]
for i in range(len(languages)):
  language = languages[i]
  print(i+1, "番目の言語は", language, "です")
```

 ### <font color=#0e89b2>◆練習問題</font>

(1) ０から99までの数字が要素として含まれたリストを作成してください。
* まず空のリストを作成します。
* 空のリストに、0から99までの数字を順番に`append()`していきます。

(2) for文を用いて、100の階乗（1×2×3×・・・×99×100）を計算してください。<br>
* まず変数`result`を定義し、仮で1を格納しておきます。
* for文を使って、`result`に順番に1から100までの数字をかけていきます。

頭5桁が93326になっていればおそらく正解です。

解答

(1)

```python
ls = []
for i in range(100):
  ls.append(i)
ls
```

(2)

```python
result = 1
for i in range(1, 101):
  result *= i
result
```

 ### <font color=#0e89b2>◆実装</font>

それでは、1,000人分のエージェントの立ち位置（x座標とy座標）をランダムに決めて、<br>
街の上に1,000人のエージェントを描画しましょう。
* 最終的には`plt.scatter([x座標のリスト], [y座標のリスト])`の形に持っていければ描画ができます。
* そこから逆算します。まずは空のx座標のリスト`x_ls`と空のy座標のリスト`y_ls`を作成します。
* `x_ls`と`y_ls`に、ランダムに決定したx座標とy座標をそれぞれ`append`します。この処理をfor文を使って1,000回繰り返します。

実装例

```python
x_ls = []
y_ls = []
for _ in range(population):
  x = np.random.randint(grid_size)
  y = np.random.randint(grid_size)
  x_ls.append(x)
  y_ls.append(y)
```

```python
plt.figure(figsize=(6,6))
plt.xlim(0, grid_size)
plt.ylim(0, grid_size)
plt.scatter(x_ls, y_ls)
```

1,000人のエージェントの描画が完了しました。<br>
おまけで、この図ではドットのサイズが大きくエージェント同士が重なりあっている箇所が多いので、<br>
ドットのサイズを小さくしてみます。<br>
<br>
ドットのサイズは、`plt.scatter()`の括弧内で、`s=10`のように指定すると調整できます。

```python
plt.figure(figsize=(6,6))
plt.xlim(0, grid_size)
plt.ylim(0, grid_size)
plt.scatter(x_ls, y_ls, s=10)
```

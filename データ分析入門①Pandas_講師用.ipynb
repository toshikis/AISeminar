## <font color=#0e89b2>2.3 乱数</font>

### <font color=#0e89b2>(1) 乱数の発生</font>

先ほど登場した`random`モジュールを使うと、色々な乱数を生成することが出来ます。<br>
例えば`randn()`関数を使うと、正規分布（綺麗な富士山型）に従う乱数を作ることが出来ます。

```python
#####写経#####
#正規分布(平均0、分散1)の乱数を10個生成
random.randn(10)
```

`rand()`関数を使うと、一様分布に従う乱数を作れます。<br>
一様分布とは、ある区間内のすべての値が等しい確率で出現する分布のことを指します。

```python
#####写経#####
#一様分布(0から1の値が等確率）の乱数を10個発生
random.rand(10)
```

他にも二項分布、ガンマ分布など、様々な種類の乱数を発生させることが出来ます。<br>
その時の状況に応じて、必要な種類のものを選びましょう。

### <font color=#0e89b2>(2) 乱数のシード</font>

上の「乱数の発生」のコードを複数回実行してみてください。<br>
当たり前ではありますが、毎回実行結果が変わるはずです。（「ランダム」なので）<br>
<br>
しかし毎回実行結果が変わってしまうと、困る場面もあります。<br>
例えば社内会議の場面では、事前の実験のときと全く同じ計算結果を見せられた方が説明がスムーズです。<br>
<br>
そこでシードという値を設定すると、何度実行しても同じ乱数（）を得ることが出来ます。

```python
#####写経#####
#シードを設定
random.seed(0) #seedの値は何でも構いません。seedの値を変えると、発生する乱数の値が変わります

#正規分布(平均0、分散1)の乱数を10個生成
random.randn(10)
```

### <font color=#0e89b2>(8) データフレーム同士の結合</font>

#### <font color=#0e89b2>a. 結合について</font>

DataFrameの結合をする場面は、実務でも多く発生します。<br>
例えば顧客管理用のcsvデータが月別に管理されている場合に、<br>
5月分のcsvデータと6月分のcsvデータをそれぞれDataFrameで読み込んで、結合するような作業が度々発生します。<br>
<br>
そこで、DataFrame同士の結合操作を練習していきましょう。<br>
<br>
なお余談として、上記のような結合作業を行う際には、以下のようなデータ処理上の様々なトラブルが発生します。

*   5月のデータに存在するカラムが6月はなくなっている
*   5月のカラム名は「社員数」なのに6月は「従業員数」に変わっている
*   売上の単位が万円から百万円に変わっている
*   データの欠損箇所が大量にある

このような事態に対処するためにも、Pandasの様々な操作を練習していく必要があります。

#### <font color=#0e89b2>b. Pandasの3種類の結合方法</font>

Pandasには3種類の結合メソッドが用意されています。<br>

◆3種類の結合メソッドの使い分け

---

*   `concat`：シンプルに縦or横に結合する
*   `merge`：特定の列をキーにして結合する
*   `join`：特定の行（インデックス）をキーにして結合する

---

<br>

抽象的な説明をするよりも具体例を見るのが早いと思うので、それぞれ見ていきましょう。

##### <font color=#0e89b2>① concatを使った結合</font>

concatは、同じ構造を持ったデータフレーム同士を結合するときに使います。<br>
まず、同じ構造を持ったデータフレームを2つ用意してみます。

```python
#####写経#####
#1個目のデータ
data1 = DataFrame({"Country":["Japan","America","Korea"],
                     "Age":[18,21,23],})
data1
```

```python
#####写経#####
#2個目のデータ
data2 = DataFrame({"Country":["Australia","America","UK"],
                     "Age":[20,25,22],})
data2
```

これらをconcatを使って結合してみます。<br>
<br>
`pd.concat(["データフレーム名1", "データフレーム2"])`

```python
#####写経#####
#二つのデータフレームを結合
new_data = pd.concat([data1, data2])
new_data
```

上手く結合できたと思います。<br>
ただし、インデックスが重複してしまいました。<br>
インデックスを振り直してみましょう。<br>
<br>
`データフレーム名.reset_index()`<br>
引数に`drop=True`を指定すると、元のインデックスが削除されます。

```python
#####写経#####
#インデックスを振り直し
new_data.reset_index(drop=True)
```

##### <font color=#0e89b2>② mergeを使った結合</font>

`merge`は、特定の列をキーにして結合するときに使います。<br>
早速例を見てみましょう。<br>
データを準備します。

```python
#####写経#####
#1個目のデータ
data1 = DataFrame({"Name":["Mizuki", "David", "Jack"],
                     "Country":["Japan","America","Australia"],
                     "Age":[18,21,23],})
data1
```

```python
#####写経#####
#2個目のデータ
data2 = DataFrame({"Name":["Mizuki", "David", "Jack"],
                     "sex":["female", "male", "male"]})
data2
```

この2つのデータを、Nameをキーにして結合してみましょう。<br>
<br>
`pd.merge(データフレーム1, データフレーム2, on="キーにする列名")`

```python
#####写経#####
#2つのデータフレームを結合
pd.merge(data1, data2, on="Name")
```

##### <font color=#0e89b2>③ joinを使った結合</font>

`join`は、特定の行（インデックス）をキーにして結合する場合に使用します。<br>
<br>
`データフレーム1.join(データフレーム2)`<br>
<br>
ちなみに、mergeのときのような、下記のような書き方はjoinではできません。<br>
`pd.join(データフレーム1, データフレーム2)`

```python
#2つのデータフレームを結合（エラーが出ます）
#data1.join(data2)
```

また、上の例だとエラーが出てしまうはずです。<br>
これは、"Name"のカラムが重複してしまっており、同名のカラムが2つ出来てしまうためです。<br>
このような場合には、下記のように引数に`lsuffix`を指定することで新旧のカラムに別の名前が命名され、解決できます。<br>
どうなるのか見てみるのが早いでしょう。

```python
#####写経#####
#2つのデータフレームを結合
data1.join(data2, lsuffix="_old")
```

#### <font color=#0e89b2>c. 結合の4パターン</font>

ここまでデータ結合の「方法(how)」を3種類見てきましたが、<br>
それと組み合わせて以下のような4種類の結合「結果」を選ぶことが出来ます。<br>
<br>
下の図で、4種類の結合パターンをご覧ください。
ちなみにここまで行ってきた結合は「内部結合」といい、データ1とデータ2両方に共通する値「キー」があるデータのみ残して結合するものです。


3183F779-960C-4FF3-8B9F-177F8E59C91D.jpeg

具体的な例で見てみましょう。<br>
<br>
データの変化をわかりやすくするため、ここからは下記のデータを用います。

```python
#データ1
data1 = {"ID":["00","01","02","03","04","06","07","10","12","13"],
               "Grade":["1","3","4","1","2","6","3","5","2","6"],
               "Score":[95,97,85,90,83,74,86,93,73,87],
               "Sex":["male","female","female","male","male","female","male","male","female","male"] }

data1 = DataFrame(data1)

data1
```

```python
#データ2
data2 = {"ID":["00","01","05","06","08","09","11","13"],
               "Grade":["1","3","2","6","4","5","4","6"],
               "Name":["Asahi","Yuika","Takeru","Rei","Hikaru","Nami","Hikari","Haruto"],
               "Class":["3","2","3","1","4","3","3","2"]}

data2 = DataFrame(data2)

data2
```

##### <font color=#0e89b2>① 内部結合</font>

内部結合では下記のようにonパラメータでidをキーとして結合を行います。

```python
#####写経#####
#内部結合のデータのマージ
pd.merge(data1, data2, on="ID")
```

##### <font color=#0e89b2>② 全結合</font>

全結合では下記のようにhowパラメータでouterをキーとして結合を行います。結合する値がない場合にはNaNになります

```python
#####写経#####
#全結合のデータのマージ
pd.merge(data1, data2, how="outer")
```

##### <font color=#0e89b2>③ 左外部結合</font>

左外部結合はhowパラメータにleftを指定します。


```python
#####写経#####
#左外部結合のデータのマージ
pd.merge(data1, data2, how="left")
```

##### <font color=#0e89b2>④ 右外部結合</font>

右外部結合はhowパラメータにrightを指定します。

```python
#####写経#####
#右外部結合のデータのマージ
pd.merge(data1, data2, how="right")
```

#### <font color=#0e89b2>b. グループ集計</font>

`groupby`メソッドを用いると、グループ単位で集計を行うことができます。<br>
例えば、クラス別に集計したい、都道府県別に集計したいといった場面で活用できます。<br>
具体的には、以下のような集計を行うことが出来ます。<br>
<br>


---
* sum()：各グループの合計値を計算します。
* mean()：各グループの平均値を計算します。
* std()：各グループの標準偏差を計算します。
* count()：各グループの行数を計算します。
* size()：各グループの行数を計算します（count()とは異なり、欠損値もカウントします）。
* max()：各グループの最大値を計算します。
* min()：各グループの最小値を計算します。
* describe()：各グループの基本統計量（count、mean、std、min、25%、50%、75%、max）を計算します。
* agg(func)：funcで指定した関数を各グループに適用します。
* transform(func)：funcで指定した関数を各グループに適用し、元のデータフレームのサイズとインデックスを保持します。
* filter(func)：funcで指定した条件を満たすグループだけを抽出します。


---


実際にコードで色々な集計例を見てみます。

##### <font color=#0e89b2>① 各グループのデータ数の集計</font>

```python
#####写経#####
#性別ごとの人数を集計
data1.groupby("Sex").size()
```

```python
#####写経#####
#性別ごとの英語の平均点を集計
data1.groupby("Sex")["English_Score"].mean()
```

##### <font color=#0e89b2>③ 各グループの平均値を集計し、平均値が高い方のグループを表示</font>

```python
#####写経#####
#性別ごとの数学の平均点を集計し、平均点が高い方の性別を表示
#1行で書いてしまっていますが、理解が追いつかない場合は分解して一つ一つの実行結果を確認してみましょう。
data1.groupby("Sex")["Math_Score"].mean().idxmax()
```

##### <font color=#0e89b2>④ 集計結果をDataFrame形式でまとめて表示</font>

```python
#####写経#####
#aggメソッドを使用して、性別×科目ごとの平均点を集計（実行結果がDataFrame形式で返ってくるので扱いやすい）
data1.groupby("Sex").agg({"Math_Score": "mean", "English_Score": "mean"})
```

##### <font color=#0e89b2>⑤ 集計結果をDataFrame形式でまとめて表示（変数ごとに複数の集計方法を適用し表示）</font>

```python
#####写経#####
#aggメソッドを使用して、複数の集計結果をまとめて表示
data1.groupby("Sex").agg({"Math_Score": ["sum", "mean"], "English_Score": ["sum", "mean"]})
```

*   システムが故障で停止していて欠損になっている

このような様々な要因で発生した欠損値や異常値に対して、<br>

#### <font color=#0e89b2>b. 欠損値処理のパターン</font>

欠損値処理のやり方には、大きく2パターンあります。<br>
「削除する」か「穴埋めする」かです。<br>
もう少し丁寧にまとめると、以下のようになります。<br>
<br>

◆欠損値処理の方法

---

*   ①削除する（データ数が少なくなる恐れがあるが、バイアスは生みにくい）
 * リストワイズ削除：全ての列にデータがある行だけを残す
 * ペアワイズ削除：必要な列(変数)だけで見て、全ての列にデータがある行だけを残す
*   ②穴埋めする（バイアスを生む恐れがあるが、データは減らない）
 * 平均値で埋める
 * 0で埋める
 * 前の値で埋める
 * その他（最尤推定法やスプライン補間で求めた値で埋める）

---

#### <font color=#0e89b2>b. ビン分割</font>

ビン分割は、範囲ごとにデータにラベル付けをする処理です。<br>
<br>
例えば学校の成績で、こんなラベル付けがありませんでしたか？<br>

*   80点以上：優
*   60〜80点：良
*   40〜60点：可
*   40点以下：不可

Pandasでは、cut関数またはqcut関数を使うとビン分割を行うことができます。<br>
<br>



---
*   `cut()`関数：等間隔または任意の境界値でビン分割
 * 例）テストの点数を80、85、90、95,100のように5点単位で分割
*   `qcut()`関数：ビンに含まれる要素数が等しくなるようにビン分割
 * 例）テストの成績上位5名まで、10名まで、15名までのように、サンプル数が5人ずつになるように分割
---

<br>

今回はcut関数の方をコーディングしてみましょう。


```python
#####写経#####
#データ
data = [["経済基礎",52],["データ分析",74],["フランス語文法",33],["社会心理学",77]]
data = pd.DataFrame(data, columns=["Subject","Score"])
data
```

```python
#####写経#####
#分割する範囲
score_bins = [0,40,60,80,100]

#ラベル名
group_names = ["不可","可","良","優"]

#ビン分割
#該当する出たがないときはNaNと表示されます
data["Grade"] = pd.cut(data.Score,score_bins,labels=group_names)
data
```

ちなみに、上の分割範囲は、デフォルトでは「第1引数<X=<第2引数」となっています。<br>
この区切りを「未満」にするか「以下」にするかなどの設定は、<br>
cut関数にmleftオプションやrightオプションを指定することで変更できます。

#### <font color=#0e89b2>e. ダミーコーディング</font>

##### <font color=#0e89b2>① 多重共線性とは</font>

上のOne-Hotエンコーディングでは、ある問題があります。<br>
それは、「多重共線性」が発生してしまうことです。<br>
<br>
多重共線性とは、一般的に以下のように説明されます。

---

複数の説明変数間に高い相関関係がある場合に、回帰係数の推定値が不安定になり、重回帰分析モデル等の精度が低下してしまうこと

---

<br>

しかしこの説明では難しいので、もう少し噛み砕いて見ていきます。<br>
例えば上の性別の例だと、「Sex_女」「Sex_男」のカラムは、表裏の関係に過ぎず、実質的には全く同じ情報を表しています。<br>
つまり、2つのカラムの情報が重複してしまっている状態です。<br>
<br>
ところが重回帰などの機械学習モデルでは、情報が重複したデータを学習させることはあまり好ましくありません。<br>
<br>
なぜならば、これらのモデルは、<br>
「どの変数が大事か、どの変数があまり大事じゃないか」<br>
を判断してそれを数値化しているわけですが、<br>
情報が重複したデータを与えると、この判断をモデルが迷ってしまうからです。

##### <font color=#0e89b2>② ダミーコーディングとは</font>

そこで上のような情報の重複を取り除いてカテゴリ変数を変換する方法が、ダミーコーディングです。<br>
方法としてはシンプルで、上の例では「Sex_女」か「Sex_男」のいずれかのカラムを削除してしまうだけです。<br>
<br>
ちなみに今回は女か男かの2値でしたが、例えばA組B組C組のような3値になっても、いずれか1つのカラムを削除すればOKです。<br>
理由は考えてみればすぐわかると思うので、ここでは説明は割愛します。<br>
<br>
例によって実装していきます。まずはデータの準備から。





```python
#データの準備
data = [["Keisuke","男",22],["Risa","女",24],["Miho","女",27],["Toshi","男",28]]
data = pd.DataFrame(data, columns=["Name","Sex","Age"])
data
```

One-Hotエンコーディングの時と同様、Pandasの`get_dummies()`関数を使います。<br>
One-Hotエンコーディングの時との違いは、引数に`drop_first=True`を指定するだけです。

```python
#####写経#####
#ダミーコーディング
pd.get_dummies(data, columns=['Sex'],drop_first=True)
```

ここまで、3種類のカテゴリ変数処理方法を見てきました。<br>
<br>
ダミーエンコーディングはOne-Hotエンコーディングの上位互換と言っても良いので、<br>
実質ラベルエンコーディングとダミーエンコーディングの2種類でしょうか。<br>
<br>
実際に使用する際には、それぞれのやり方に一長一短あるので、状況に応じて適した方を使うのが良いです。<br>
<br>
以下に各方法の主なデメリットを挙げていきます。<br>
例えば使用する機械学習モデルの種類によっては、ここで記載しているデメリットが無関係だったりもするので、<br>
適した方を選びましょう。
<br>

---
◆ラベルエンコーディング：
 * 意味のない数値の大小が表現されてしまう
  * 例えば都道府県をラベル付けして、北海道を0、沖縄を46とした時に、そこに本来大小関係はないはずでしたが、あたかも数値上大小関係があるように表現されてしまいます

◆ダミーエンコーディング：
 * カテゴリ数がたくさんあるデータでは、カラムが大量に生成されてしまう
  * 例えば都道府県のデータでは、46列分のカラムが生成されてしまいます
  * このようにデータ量が増えると、メモリを食って処理が遅くなるなどの問題が発生します
 * ラベル間の序列を表現できなくなる
  * 例えば学歴のデータでは、一般に高卒よりも大卒の方が序列は上の扱いになりますが、この序列を表現できなくなります

### <font color=#0e89b2>総合問題2-2</font>

一般的に<font color=#f27405>「でかい」モンスターは「強い」イメージ</font>があります。<br>
しかしポケモンにはエスパータイプなどもいるので、小柄でも強いポケモンもいます。

image.png

それではポケモンでは、<font color=#f27405>「でかさ」と「強さ」の間にどれくらいの相関</font>があるのでしょうか？

(1) 「でかさ」を表すデータが欲しいところです。<br>
"height"のデータはありますが、例えば細長いだけで「でかい」感じはしないポケモンなどもいるため、<br>
"height"だけでは「でかさ」を表現するには不十分そうです。<br>
<br>
そこで"height"と"weight"を組み合わせて、「でかさ」を表す新たな変数を作ってみてください。<br>
* "height"を5階級に分けて、1〜5（体長が大きい方が5）までのランク値を振ってください
 * 階級分けの際には、データ個数が均等になるように5つに分けてください
* "weight"を5階級に分けて、1〜5（体重が重い方が5）までのランク値を振ってください
 * 階級分けの際には、データ個数が均等になるように5つに分けてください
* "height"のランク値と"weight"のランク値を掛け合わせて、「でかさ」を表す変数を作ってください（取り得る値は1〜25）

image.png

```python
#「でかさ」を表す新たな変数の作成
```

解答

```python
#「でかさ」を表す新たな変数の作成
pokemon['height_size'] = pd.qcut(pokemon['height_m'], 5, labels=['1', '2', '3', '4', '5']).astype(float)
pokemon['weight_size'] = pd.qcut(pokemon['weight_kg'], 5, labels=['1', '2', '3', '4', '5']).astype(float)
pokemon['size'] = pokemon['height_size'] * pokemon['weight_size']
pokemon.head()
```

(2) (1)で作ったカテゴリ変数と総合能力値の相関を求めてください。

```python
#「でかさ」と総合能力値の相関
```

解答

```python
#「でかさ」と総合能力値の相関
pokemon['total'].corr(pokemon['size'])
```

```python
#「でかさ」と相関の低い能力値
```
